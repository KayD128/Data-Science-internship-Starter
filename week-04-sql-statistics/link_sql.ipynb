```python
from sqlalchemy import text

create_table = text("""
                    CREATE TABLE employees (
                    employee_id INT PRIMARY KEY, 
                    name VARCHAR(100), 
                    department VARCHAR(50), 
                    salary INT, 
                    years_experience INT, 
                    hire_date DATE
                    );
                    """)

with engine.connect() as conn:
    conn.execute(create_table)
    conn.commit()
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)

conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

cursor.execute('''CREATE TABLE sales (
    sale_id INTEGER PRIMARY KEY,
    product_name VARCHAR(100),
    department_id INTEGER,
    quantity INTEGER,
    price DECIMAL(10,2),
    sale_date DATE,
    customer_name VARCHAR(100),
    FOREIGN KEY (department_id) REFERENCES department(department_id)
);''')

cursor.execute('''
CREATE TABLE department (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(100),
    manager_name VARCHAR(100),
    location VARCHAR(100)
);''')
```

```python
department_data = [
    (1,	"Electronics", "Sarah", "Lagos"),
    (2,	"Clothing", "David", "Abuja"),
    (3,	"Groceries", "Mary", "Port Harcourt"),
    (4,	"vehicles", "James", "Ibadan"),
]

sales_data = [
    (1, "Laptop", 1, 1200, "2025-03-01", "John", 1),
    (2, "Shirt", 3, 25, "2025-03-02", "Mary", 2),
    (3, "Rice", 5, 10, "2025-03-02", "David", 3),
    (4, "Corolla", 1, 10, "2025-03-03", "Micheal", 4),
    (5, "Garri", 5, 10, "2025-03-13", "Kolade", 3),
    (6, "Laptop", 2, 10, "2025-03-06", "Victor", 1),
    (7, "Beans", 3, 10, "2025-03-08", "Treasure", 3),
]

cursor.executemany('INSERT INTO department VALUES (?,?,?,?)', department_data)
cursor.executemany('INSERT INTO sales VALUES (?,?,?,?,?,?,?)', sales_data)
```
